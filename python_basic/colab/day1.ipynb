## 📚 핵심 개념

### 1️⃣ 변수란? — 상자에 이름표 붙이기

변수는 **값을 담는 상자**예요. 상자에 이름표를 붙여두면 나중에 이름만 불러서 값을 꺼낼 수 있어요.

`이름 = "김파이"` 라고 쓰면, "김파이"라는 값을 담은 상자에 "이름"이라는 이름표를 붙인 거예요.

### 2️⃣ 자료형 — 상자 안에 뭐가 들었을까?

파이썬의 기본 자료형 4가지를 알아두세요.

- **int** (정수): 17, -3, 0 — 소수점 없는 숫자

- **float** (실수): 3.14, 165.5 — 소수점 있는 숫자

- **str** (문자열): "안녕", '파이썬' — 따옴표로 감싼 글자

- **bool** (불리언): True, False — 참/거짓 (첫 글자 대문자!)

### 3️⃣ print() — 화면에 보여주기

`print()`는 괄호 안의 내용을 화면에 출력해요. 쉼표로 여러 개를 한 번에 출력할 수 있어요.

### 4️⃣ input() — 사용자에게 물어보기

`input()`은 키보드로 입력받은 값을 돌려줘요. **⚠️ 주의: input()으로 받은 값은 무조건 문자열(str)이에요!** 숫자로 계산하려면 `int()`로 바꿔줘야 해요.

### 5️⃣ 연산자 — 계산기 놀이

`+` `-` `*` `/` 외에 파이썬만의 특별한 연산자가 있어요.

- `//` : 몫만 (소수점 버림) → 7 // 3 = 2

- `%` : 나머지만 → 7 % 3 = 1

- `**` : 거듭제곱 → 2 ** 3 = 8

### 6️⃣ f-string — 문자열 속에 변수 넣기

문자열 앞에 `f`를 붙이고, 변수 자리에 `{중괄호}`를 쓰면 값이 쏙 들어가요. 가장 편한 방법이에요!

## 💻 따라 치기 코드

### ① 변수 만들기

```python
# 1. 변수란? 상자에 이름표를 붙이는 것!

# 상자(이름표 = 값) 만들기
이름 = "김파이"      # 문자열 상자
나이 = 17           # 정수 상자
키 = 165.5         # 실수 상자
학생인가요 = True   # 불리언 상자 (참/거짓)

# 상자 안을 들여다보기
print(이름)
print(나이)
print(키)
print(학생인가요)
```

### ② 자료형 확인하기

```python
# 2. 자료형 알아보기 - type()으로 확인!
print(type(17))        # 정수 int
print(type(3.14))      # 실수 float
print(type("안녕"))    # 문자열 str
print(type(True))      # 불리언 bool

# 변수에 담긴 값의 자료형도 확인할 수 있어요
나이 = 17
print(type(나이))      # 정수 int
```

### ③ print()로 출력하기

```python
# 3. print()로 출력하기
print("안녕하세요!")          # 그냥 출력

# 쉼표로 여러 개 출력 (자동으로 띄어쓰기가 들어가요)
print("나는", 17, "살이야")

# 문자열끼리 더하기 (붙여쓰기)
이름 = "김파이"
print("내 이름은 " + 이름 + "야")
```

### ④ input()으로 입력받기

```python
# 4. input()으로 입력받기
# ⚠️ input()으로 받은 값은 무조건 문자열(str)이에요!

이름 = input("이름을 입력하세요: ")
print("반가워요,", 이름, "님!")

# 숫자로 계산하려면 int()로 바꿔줘야 해요
나이 = input("나이를 입력하세요: ")
나이 = int(나이)  # 문자열 → 정수로 변환
print("내년 나이는", 나이 + 1, "살이네요!")
```

### ⑤ 연산자

```python
# 5. 연산자 - 계산기 놀이!
print(7 + 3)    # 10  더하기
print(7 - 3)    # 4   빼기
print(7 * 3)    # 21  곱하기
print(7 / 3)    # 2.333... 나누기 (소수점까지)
print(7 // 3)   # 2   몫만 (소수점 버림)
print(7 % 3)    # 1   나머지만
print(2 ** 3)   # 8   거듭제곱 (2의 3제곱)

# 변수에 담아서 계산해도 돼요
가격 = 1500
개수 = 3
합계 = 가격 * 개수
print("합계:", 합계)
```

### ⑥ f-string

```python
# 6. f-string - 문자열 속에 변수 넣기 (제일 편한 방법!)
이름 = "김파이"
나이 = 17

# f를 앞에 붙이고, 변수 자리에 {중괄호}!
print(f"내 이름은 {이름}이고, 나이는 {나이}살이야.")

# 계산식도 바로 넣을 수 있어요!
print(f"내년 나이는 {나이 + 1}살이야.")

# 소수점 자릿수도 조절할 수 있어요
평균 = 87.66666
print(f"평균 점수는 {평균:.1f}점이야.")  # 소수점 첫째 자리까지
```
