## 📚 핵심 개념

### 1️⃣ if / elif / else — 갈림길에서 선택하기

교통카드 잔액이 1500원 이상이면 버스를 타고, 아니면 충전하러 가죠? 프로그램도 이렇게 **조건에 따라 다른 행동**을 해요.

`if` (만약 ~라면) → `elif` (아니면 만약 ~라면) → `else` (그 외에는) 순서로 써요.

⚠️ **들여쓰기**가 생명이에요! 조건 아래 실행할 코드는 스페이스 4칸(또는 Tab 1번) 들여쓰기!

### 2️⃣ 비교 연산자 — 두 값을 비교해요

⚠️ `=` 하나는 "넣어라(대입)", `==` 두 개는 "같니?(비교)"예요. 헷갈리면 큰일나요!

### 3️⃣ 논리 연산자 — 조건을 조합해요

- `and`: 둘 다 참이어야 참 (그리고)

- `or`: 하나라도 참이면 참 (또는)

- `not`: 참↔거짓 뒤집기 (아니다)

예: `나이 >= 17 and 학생증` → "17살 이상 **그리고** 학생증이 있을 때"

### 4️⃣ for문 — 정해진 횟수만큼 반복

`range(5)`는 0부터 4까지 5개의 숫자를 만들어요. `for i in range(5):`는 "5번 반복해라"는 뜻!

- `range(5)` → 0, 1, 2, 3, 4

- `range(1, 6)` → 1, 2, 3, 4, 5 (끝 숫자는 포함 안 돼요!)

- `range(2, 11, 2)` → 2, 4, 6, 8, 10 (2씩 뛰기)

### 5️⃣ while문 — 조건이 참인 동안 계속

`while 조건:`은 조건이 참인 동안 계속 반복해요. **⚠️ 조건이 영원히 참이면 무한 반복!** 반복 안에서 조건을 바꿔주는 코드를 꼭 넣으세요.

### 6️⃣ break와 continue — 반복문 조종하기

- `break`: 반복문을 **완전히 탈출!**

- `continue`: 이번 차례만 **건너뛰고** 다음으로!

## 💻 따라 치기 코드

### ① if / elif / else

```python
# 1. if문 - 교통카드 잔액 비유!
# 잔액이 1500원 이상이면? 버스를 탈 수 있어요!

잔액 = 2000

if 잔액 >= 1500:
    print("🚌 버스에 탑승합니다!")
else:
    print("💳 잔액이 부족해요. 충전하세요!")

print()

# if / elif / else - 여러 갈래 길
점수 = 85

if 점수 >= 90:
    print("A학점! 🎉")
elif 점수 >= 80:
    print("B학점! 😊")
elif 점수 >= 70:
    print("C학점! 🙂")
else:
    print("조금 더 노력해요! 💪")

# ⚠️ 들여쓰기를 꼭 지키세요! (스페이스 4칸)
```

### ② 비교 연산자

```python
# 2. 비교 연산자 - 두 값을 비교해요 (결과는 True/False!)
print(5 == 5)   # True  (같다)
print(5 != 5)   # False (다르다)
print(5 > 3)    # True  (크다)
print(5 < 3)    # False (작다)
print(5 >= 5)   # True  (크거나 같다)
print(5 <= 4)   # False (작거나 같다)

print()

# ⚠️ = 하나는 "넣어라(대입)", == 두 개는 "같니?(비교)"
나이 = 17          # 17을 나이 상자에 넣기
print(나이 == 17)  # 나이가 17과 같니? → True
print(나이 == 20)  # 나이가 20과 같니? → False
```

### ③ 논리 연산자

```python
# 3. 논리 연산자 - 조건을 조합해요!
# and: 둘 다 참이어야 참 (그리고)
# or: 하나라도 참이면 참 (또는)
# not: 뒤집기 (아니다)

나이 = 17
학생증 = True

# and: 나이도 17 이상 "그리고" 학생증도 있어야
if 나이 >= 17 and 학생증:
    print("영화관 청소년 할인! 🎬")

# or: 비가 오"거나" 눈이 오면
날씨 = "맑음"
if 날씨 == "비" or 날씨 == "눈":
    print("우산을 챙기세요! ☂️")
else:
    print("좋은 날씨네요! ☀️")

# not: 뒤집기
로그인 = False
if not 로그인:
    print("로그인이 필요합니다. 🔑")
```

### ④ for문과 range()

```python
# 4. for문 - 정해진 횟수만큼 반복!
# range(5) → 0, 1, 2, 3, 4 (5번 반복)

for i in range(5):
    print(f"{i+1}번째 안녕! 👋")

print()

# range(시작, 끝) → 시작부터 끝-1까지
for 숫자 in range(1, 6):
    print(숫자)  # 1, 2, 3, 4, 5

print()

# range(시작, 끝, 간격) → 2씩 뛰기
print("짝수만 출력:")
for n in range(2, 11, 2):
    print(n)  # 2, 4, 6, 8, 10
```

### ⑤ while문

```python
# 5. while문 - 조건이 참인 동안 계속 반복!
# ⚠️ 조심! 조건이 영원히 참이면 무한 반복이에요

횟수 = 1
while 횟수 <= 5:
    print(f"{횟수}번째 반복 중...")
    횟수 = 횟수 + 1  # ← 이걸 빼먹으면 무한 루프! 😱

print("반복 끝! 🎉")
```

### ⑥ break와 continue

```python
# 6. break와 continue - 반복문 조종하기!
# break: 반복문을 완전히 탈출!
# continue: 이번 차례만 건너뛰고 다음으로!

print("--- break 예시 ---")
for i in range(1, 11):
    if i == 4:
        print("4에서 멈춤! 🛑")
        break  # 반복문 탈출!
    print(i)

print()
print("--- continue 예시 ---")
for i in range(1, 6):
    if i == 3:
        continue  # 3만 건너뛰기
    print(i)  # 1, 2, 4, 5 출력 (3은 안 나와요)
```

## 🛠️ 실습 1: 숫자 맞추기 게임

컴퓨터가 정한 비밀 숫자를 맞추는 게임! `random` 모듈을 사용해요.

```python
# 🎯 숫자 맞추기 게임!
import random  # 랜덤 숫자 모듈 불러오기

# 1부터 20 사이의 비밀 숫자 정하기
비밀숫자 = random.randint(1, 20)
시도횟수 = 0

print("🎯 숫자 맞추기 게임!")
print("1부터 20 사이의 숫자를 맞춰보세요.")
print()

while True:  # 맞출 때까지 무한 반복
    추측 = int(input("숫자를 입력하세요: "))
    시도횟수 = 시도횟수 + 1

    if 추측 == 비밀숫자:
        print(f"🎉 정답! {시도횟수}번 만에 맞췄어요!")
        break  # 맞췄으니 탈출!
    elif 추측 < 비밀숫자:
        print("⬆️ 더 큰 숫자예요!")
    else:
        print("⬇️ 더 작은 숫자예요!")
```

## 🛠️ 실습 2: 구구단 출력

for문 하나로 구구단 완성! 원하는 단을 입력받아보세요.

```python
# ✖️ 구구단 출력하기
단 = int(input("몇 단을 출력할까요? "))

print(f"--- {단}단 ---")
for i in range(1, 10):
    print(f"{단} × {i} = {단 * i}")

print()
print("--- 도전! 구구단 전체 출력 ---")
for 단 in range(2, 10):
    print(f"[{단}단]", end="  ")
    for i in range(1, 10):
        print(f"{단*i:2}", end=" ")
    print()  # 줄바꿈
```
