## 📁 1. 파일 읽기/쓰기 — 메모장에 저장하기

### 파일이란?

지금까지 만든 데이터는 프로그램이 끝나면 **사라집니다.** 파일에 저장하면 **영구 보관**됩니다!

- `"w"` — 쓰기 모드 (새로 만들기, 기존 내용은 지워짐!)

- `"r"` — 읽기 모드

- `"a"` — 이어쓰기 모드 (기존 내용 뒤에 추가)

**with문**을 쓰면 파일을 다 쓴 뒤 **자동으로 닫아줍니다.** (닫는 걸 깜빡해도 안전!)

```python
# 파일에 쓰기 (w = 새로 쓰기)
with open("일기.txt", "w", encoding="utf-8") as f:
    f.write("오늘 파이썬 4일차를 배웠다!\n")  # \n = 줄바꿈
    f.write("파일 저장이 신기하다.\n")
# ← with 블록이 끝나면 자동으로 파일 닫힘!

# 파일 읽기 (r = 읽기)
with open("일기.txt", "r", encoding="utf-8") as f:
    내용 = f.read()      # 전체 내용을 한 번에 읽기
    print(내용)

# 이어쓰기 (a = 뒤에 추가)
with open("일기.txt", "a", encoding="utf-8") as f:
    f.write("내일은 5일차다!\n")

# 다시 읽어보기
with open("일기.txt", "r", encoding="utf-8") as f:
    print("=== 최종 내용 ===")
    print(f.read())
```

## 🧩 2. 모듈과 import — 남의 코드를 빌려 쓰기

### 모듈이란?

다른 사람이 미리 만들어둔 **코드 묶음(도구상자)**입니다. `import`로 가져와서 씁니다.

비유: **레고 블록** — 하나하나 만들지 말고 완성된 블록을 조립하세요!

### 🎲 random — 랜덤 도구

```python
import random  # 랜덤 도구 가져오기!

# 주사위 굴리기 (1~6 중 하나)
print(random.randint(1, 6))

# 리스트에서 랜덤으로 하나 뽑기
점심메뉴 = ["김밥", "라면", "떡볶이", "비빔밥"]
print(random.choice(점심메뉴))

# 순서 섞기
카드 = ["A", "B", "C", "D"]
random.shuffle(카드)
print(카드)
```

### 📅 datetime — 날짜·시간 도구

```python
import datetime  # 날짜 도구 가져오기!

# 지금 시간
지금 = datetime.datetime.now()
print(지금)

# 예쁘게 출력하기
print(지금.strftime("%Y년 %m월 %d일"))   # 2026년 10월 14일
print(지금.strftime("%H시 %M분"))       # 15시 30분

# 오늘 날짜만
오늘 = datetime.date.today()
print(오늘)
```

## 🛡️ 4. 예외 처리 — 오류가 나도 멈추지 않기

### 예외 처리란?

프로그램이 오류로 **갑자기 죽지 않게** 미리 대비하는 것입니다.

비유: **안전벨트** — 사고가 나도 최소한의 안전을 보장합니다!

- `try:` — "일단 시도해봐" (위험한 코드)

- `except:` — "오류 나면 이렇게 해" (대처 코드)

```python
# 예외 처리 없이 나누기 → 0으로 나누면 프로그램이 죽어요!
# print(10 / 0)   # ❌ ZeroDivisionError!

# try/except로 안전하게!
try:
    결과 = 10 / 0          # 위험한 시도!
    print(결과)
except:
    print("⚠️ 0으로 나눌 수 없어요!")   # 오류 나면 이쪽 실행

print("프로그램 계속 실행 중...")  # ← 죽지 않고 계속됨!

# 숫자 입력 받기 (문자를 입력하면 오류!)
try:
    숫자 = int(input("숫자를 입력하세요: "))
    print("입력한 숫자의 2배:", 숫자 * 2)
except:
    print("⚠️ 숫자를 입력해주세요!")

# 어떤 오류인지 알고 싶을 때
try:
    f = open("없는파일.txt", "r")
except FileNotFoundError:
    print("⚠️ 파일이 없습니다!")
except Exception as e:
    print("⚠️ 알 수 없는 오류:", e)
```

## 🎯 종합 실습: 가계부 프로그램

지금까지 배운 **파일 + 함수 + 예외 처리**를 합쳐서 가계부를 만들어봅시다!

```python
# ===== 가계부 프로그램 =====
# 파일에 지출 내역을 저장하고 불러옵니다!

파일명 = "가계부.txt"

# 지출 추가 함수 (파일에 이어쓰기)
def 지출추가(항목, 금액):
    try:
        with open(파일명, "a", encoding="utf-8") as f:
            f.write(항목 + "," + str(금액) + "\n")  # 쉼표로 구분해서 저장
        print("✅ 저장 완료: " + 항목 + " " + str(금액) + "원")
    except Exception as e:
        print("⚠️ 저장 실패:", e)

# 전체 내역 보기 함수
def 내역보기():
    try:
        with open(파일명, "r", encoding="utf-8") as f:
            내용 = f.read()
            if 내용 == "":
                print("📭 아직 내역이 없습니다.")
            else:
                print("📋 === 지출 내역 ===")
                print(내용)
    except FileNotFoundError:
        print("📭 가계부 파일이 아직 없습니다. 먼저 지출을 추가하세요!")

# 총 지출 합계 함수
def 총합계():
    try:
        합계 = 0
        with open(파일명, "r", encoding="utf-8") as f:
            for 줄 in f:                    # 한 줄씩 읽기
                항목, 금액 = 줄.strip().split(",")  # 쉼표로 나누기
                합계 += int(금액)           # 금액 더하기
        return 합계
    except FileNotFoundError:
        return 0

# ===== 사용해보기 =====
지출추가("점심", 8000)
지출추가("교통비", 1500)
지출추가("간식", 3000)

내역보기()
print("💰 총 지출:", 총합계(), "원")
```
