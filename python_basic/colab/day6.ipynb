## 💻 프로젝트 1: 영어 단어 퀴즈 게임

### 어떻게 동작하나요?

- 딕셔너리에 영어 단어와 뜻을 저장합니다.

- `random.shuffle()`로 문제 순서를 섞습니다.

- 하나씩 뜻을 물어보고 정답이면 점수를 올립니다.

- 마지막에 총 점수를 보여줍니다.

```python
# ==========================================
# 🎮 영어 단어 퀴즈 게임
# ==========================================
# 영어 단어를 보고 한국어 뜻을 맞히는 게임!
# 딕셔너리 + random + for문을 사용합니다.

import random  # 순서를 섞을 때 필요한 도구

# 1. 단어장 만들기 (딕셔너리: 영어 → 한국어)
words = {
    "apple": "사과",
    "banana": "바나나",
    "computer": "컴퓨터",
    "python": "파이썬",
    "school": "학교",
    "friend": "친구",
    "happy": "행복한",
    "book": "책",
    "water": "물",
    "music": "음악",
}

print("=" * 40)
print("🎮 영어 단어 퀴즈 게임을 시작합니다!")
print(f"총 {len(words)}문제가 나옵니다.")
print("=" * 40)

score = 0  # 맞힌 문제 수
total = 0  # 지금까지 푼 문제 수

# 2. 문제 순서 섞기
quiz_words = list(words.keys())  # 영어 단어만 꺼내서 리스트로
random.shuffle(quiz_words)       # 순서를 마구 섞기!

# 3. 하나씩 퀴즈 내기
for eng in quiz_words:
    total += 1  # 푼 문제 수 1 증가
    answer = input(f"[{total}] '{eng}'의 뜻은? ")

    if answer.strip() == words[eng]:  # strip(): 앞뒤 공백 제거
        print("✅ 정답!")
        score += 1  # 점수 1점 추가
    else:
        print(f"❌ 땡! 정답은 '{words[eng]}'입니다.")

# 4. 최종 결과 보여주기
print("=" * 40)
print(f"🏁 결과: {total}문제 중 {score}문제 정답!")
print(f"점수: {score * 10}점")

if score == total:
    print("🎉 완벽해요! 단어왕!")
elif score >= total / 2:
    print("👍 잘했어요!")
else:
    print("💪 조금만 더 연습해봐요!")
```

## 💻 프로젝트 2: 간단한 챗봇

### 어떻게 동작하나요?

- 딕셔너리에 키워드와 대답을 미리 저장합니다.

- `while True`로 대화를 계속 이어갑니다.

- 사용자의 말에 키워드가 들어있으면 맞는 대답을 합니다.

- "잘가"라고 하면 대화를 끝냅니다.

```python
# ==========================================
# 🤖 간단한 챗봇 "파이봇"
# ==========================================
# 사용자의 말에 대답하는 챗봇!
# if문 + while문 + 딕셔너리를 사용합니다.

# 1. 키워드 → 대답 딕셔너리 만들기
answers = {
    "안녕": "안녕하세요! 저는 파이썬 챗봇이에요 😊",
    "이름": "제 이름은 파이봇이에요!",
    "나이": "저는 올해 태어났어요. 아직 아기예요 👶",
    "좋아하는 음식": "전기를 좋아해요. 맛있어요 ⚡",
    "날씨": "창문을 열어보세요! 제가 밖은 못 봐요 😅",
    "취미": "코딩이요! 여러분이랑 노는 게 제일 재밌어요 💻",
    "잘가": "다음에 또 만나요! 👋",
}

print("=" * 40)
print("🤖 파이봇: 안녕! 나랑 대화해봐.")
print("   (그만하고 싶으면 '잘가'라고 입력해)")
print("=" * 40)

# 2. while문으로 대화를 계속 이어가기
while True:
    user = input("🙋 나: ")

    # 3. 딕셔너리에서 맞는 대답 찾기
    found = False  # 대답을 찾았는지 기억하는 변수
    for keyword in answers:
        if keyword in user:  # 키워드가 말 속에 들어있으면
            print(f"🤖 파이봇: {answers[keyword]}")
            found = True
            break  # 찾았으니 더 찾지 않고 나가기

    # 4. 아무 키워드도 없으면 모른다고 하기
    if not found:
        print("🤖 파이봇: 음... 그건 아직 몰라요. 다른 걸 물어봐주세요!")

    # 5. '잘가'라고 하면 대화 끝내기
    if "잘가" in user:
        break  # while문 탈출!

print("=" * 40)
print("🤖 파이봇: 대화가 끝났어요. 즐거웠어요!")
```

## 💻 프로젝트 3: 숫자 야구 게임

### 어떻게 동작하나요?

- `random.sample()`로 겹치지 않는 숫자 3개를 뽑습니다.

- 사용자가 숫자 3개를 입력하면 스트라이크/볼을 셉니다.

- 숫자+자리 모두 맞으면 스트라이크(S), 숫자만 맞으면 볼(B)!

- 3 스트라이크가 되면 게임 클리어!

```python
# ==========================================
# ⚾ 숫자 야구 게임
# ==========================================
# 컴퓨터가 만든 3자리 숫자를 맞히는 게임!
# random + for문 + 조건문을 사용합니다.
# S(스트라이크): 숫자도 자리도 맞음
# B(볼): 숫자는 맞는데 자리가 다름

import random  # 랜덤 숫자를 뽑을 때 필요한 도구

# 1. 컴퓨터가 1~9 중에서 서로 다른 숫자 3개 뽑기
# random.sample(범위, 개수): 겹치지 않게 뽑아줘!
numbers = random.sample(range(1, 10), 3)

print("=" * 40)
print("⚾ 숫자 야구 게임을 시작합니다!")
print("1~9 사이의 서로 다른 숫자 3개를 맞히세요.")
print("입력 예) 3 5 7  (띄어쓰기로 구분)")
print("=" * 40)

tries = 0  # 시도 횟수

# 2. 맞힐 때까지 계속 반복
while True:
    tries += 1
    guess = input(f"[{tries}번째] 숫자 3개 입력: ").split()

    # 입력이 올바른지 먼저 확인하기
    if len(guess) != 3:
        print("❗ 숫자 3개를 입력해주세요!")
        tries -= 1  # 잘못 입력한 건 횟수에서 빼기
        continue    # 다시 입력받으러 가기

    guess = [int(n) for n in guess]  # 문자를 숫자로 바꾸기

    # 3. 스트라이크와 볼 세기
    strike = 0
    ball = 0
    for i in range(3):
        if guess[i] == numbers[i]:
            strike += 1      # 숫자도 같고 자리도 같으면!
        elif guess[i] in numbers:
            ball += 1        # 숫자는 있지만 자리가 다르면!

    print(f"→ {strike}S {ball}B")

    # 4. 3 스트라이크면 게임 끝!
    if strike == 3:
        print("=" * 40)
        print(f"🎉 홈런! {tries}번 만에 맞혔어요!")
        break
```
