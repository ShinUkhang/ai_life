## 📐 3. 수학 연결 — 리스트로 함수 그래프

### 점을 촘촘히 찍으면 곡선이 돼요

x 값을 0.1 간격으로 101개 만들고, 각 x에 대한 y 값을 계산해 리스트에 담으면 매끄러운 함수 그래프가 됩니다. 2일차의 `for`문과 3일차의 `append`를 함께 써요.

```python
# 수학 연결: 리스트로 함수 그래프 그리기
import matplotlib.pyplot as plt

x = []
for i in range(-50, 51):
    x.append(i / 10)          # -5.0, -4.9, ..., 5.0  (점 101개)

y1 = []
y2 = []
for 값 in x:
    y1.append(값 ** 2)         # 이차함수 y = x²
    y2.append(2 * 값 + 1)      # 일차함수 y = 2x + 1

plt.plot(x, y1, label="y = x²")
plt.plot(x, y2, label="y = 2x + 1", linestyle="--")
plt.axhline(0, color="black", linewidth=0.8)   # x축
plt.axvline(0, color="black", linewidth=0.8)   # y축
plt.title("함수 그래프")
plt.legend()
plt.grid(True)
plt.show()

# 🚀 도전: import math 를 하고 y = math.sin(값) 그래프도 그려 보세요!
# ❓ 두 그래프가 만나는 점은 어디일까요? x² = 2x + 1 을 풀어서 확인해 봐요.
```

## 📄 5. CSV 파일 → 리스트 → 그래프

### CSV란?

**쉼표(,)로 칸을 나눈 표**예요. 엑셀·구글 시트·공공데이터포털(data.go.kr)·기상자료개방포털에서 CSV로 내려받을 수 있어요. 4일차에 배운 **파일 읽기**로 한 줄씩 꺼내 리스트에 담으면, 1~4번에서 배운 그래프를 그대로 그릴 수 있어요.

⚠️ CSV에서 꺼낸 값은 모두 **문자열**이에요. 1일차 `input()`처럼 `float()`·`int()`로 숫자로 바꿔야 그래프가 돼요.

### 7-1. 연습용 CSV 파일 만들기

```python
# 7-1. 연습용 CSV 파일 만들기 (4일차에 배운 파일 쓰기!)
# CSV = 쉼표(,)로 칸을 나눈 표. 엑셀·구글 시트에서 '다운로드 → CSV'로 만들 수 있어요.

내용 = """월,평균기온,강수량
1월,-2.0,16.8
2월,0.7,28.2
3월,6.1,36.9
4월,12.6,72.9
5월,18.2,103.6
6월,22.7,129.5
7월,25.3,414.4
8월,26.1,348.2
9월,21.6,141.5
10월,14.8,52.2
11월,7.2,51.1
12월,0.4,22.6
"""

with open("서울_날씨.csv", "w", encoding="utf-8") as f:
    f.write(내용)

print("서울_날씨.csv 파일을 만들었어요! 왼쪽 📁 폴더 아이콘을 눌러 확인해 보세요.")
```

### 7-2. CSV 읽어서 리스트에 담기

```python
# 7-2. CSV 파일 읽어서 리스트에 담기
import csv

월 = []
기온 = []
강수량 = []

with open("서울_날씨.csv", encoding="utf-8") as f:
    읽기 = csv.reader(f)
    제목줄 = next(읽기)              # 첫 줄은 열 이름
    print("열 이름:", 제목줄)
    for 줄 in 읽기:                   # 줄 = ["1월", "-2.0", "16.8"]  ← 모두 문자열!
        월.append(줄[0])
        기온.append(float(줄[1]))     # 문자 → 숫자로 바꿔서 담기
        강수량.append(float(줄[2]))

print(월)
print(기온)

가장더운 = 기온.index(max(기온))     # 가장 큰 값이 몇 번째에 있는지
print(f"가장 더운 달: {월[가장더운]} ({기온[가장더운]}°C)")
print(f"1년 강수량 합계: {sum(강수량):.1f} mm")
```

### 7-3. 리스트로 그래프 그리기

```python
# 7-3. CSV에서 꺼낸 리스트로 그래프 그리기 (7-2를 먼저 실행하세요!)
import matplotlib.pyplot as plt

fig, 칸 = plt.subplots(1, 2, figsize=(13, 4))

칸[0].plot(월, 기온, marker="o", color="tab:red")
칸[0].set_title("서울 월별 평균 기온")
칸[0].set_ylabel("기온 (°C)")
칸[0].grid(True)

칸[1].bar(월, 강수량, color="tab:blue")
칸[1].set_title("서울 월별 강수량")
칸[1].set_ylabel("강수량 (mm)")

plt.tight_layout()
plt.show()

# ❓ 비가 가장 많이 오는 달은? 그때 기온은 어떤가요?
```

### 7-4. 내 CSV 파일 올리기 (코랩 전용)

```python
# 7-4. 내 CSV 파일 올려서 읽기 (코랩 전용)
import csv
from google.colab import files

def csv_읽기(파일이름):
    """CSV 파일을 줄 리스트로 돌려줘요. 엑셀에서 저장한 한글 파일(cp949)도 읽어요."""
    for 인코딩 in ["utf-8-sig", "cp949"]:
        try:
            with open(파일이름, encoding=인코딩) as f:
                return list(csv.reader(f))
        except UnicodeDecodeError:
            pass        # 이 방식으로 못 읽으면 다음 방식으로!
    print("😢 파일을 읽지 못했어요. CSV 파일이 맞는지 확인해 주세요.")
    return []

올린파일 = files.upload()            # '파일 선택' 버튼이 나타나요

if not 올린파일:
    print("파일을 고르지 않았어요. 이 칸을 다시 실행해 주세요.")
else:
    파일이름 = list(올린파일.keys())[0]
    표 = csv_읽기(파일이름)
    print("올린 파일:", 파일이름, "/ 줄 수:", len(표))
    for 번호, 이름 in enumerate(표[0]):
        print(f"  {번호}번 열: {이름}")      # 몇 번 열을 그래프로 그릴지 고르세요
    print("앞의 3줄:", 표[1:4])
```

## 🛠️ 종합 실습: 우리 반 설문 데이터 미니 보고서

CSV 만들기 → 리스트에 담기 → 딕셔너리로 세기 → 그래프 3개 → 알게 된 점까지, 데이터 분석의 전체 흐름을 한 번에 해 봐요.

```python
# 🛠️ 종합 실습: 우리 반 설문 데이터 미니 보고서
import csv
import matplotlib.pyplot as plt

# ① 설문 결과 CSV 만들기 (실제 수업에서는 구글 설문 결과를 CSV로 받아 올려도 돼요)
내용 = """번호,통학시간,수면시간,좋아하는과목
1,15,7.0,수학
2,40,6.0,영어
3,25,6.5,과학
4,10,8.0,수학
5,55,5.5,국어
6,30,6.5,과학
7,20,7.5,수학
8,45,6.0,사회
9,35,6.5,영어
10,5,8.5,과학
11,50,5.0,수학
12,25,7.0,국어
"""
with open("설문.csv", "w", encoding="utf-8") as f:
    f.write(내용)

# ② 읽어서 리스트에 담기
통학 = []
수면 = []
과목들 = []
with open("설문.csv", encoding="utf-8") as f:
    읽기 = csv.reader(f)
    next(읽기)
    for 줄 in 읽기:
        통학.append(int(줄[1]))
        수면.append(float(줄[2]))
        과목들.append(줄[3])

# ③ 딕셔너리로 과목별 인원 세기 (3일차 복습!)
과목수 = {}
for 과목 in 과목들:
    과목수[과목] = 과목수.get(과목, 0) + 1
print("과목별 인원:", 과목수)

# ④ 그래프 3개로 보고서 만들기
fig, 칸 = plt.subplots(1, 3, figsize=(16, 4))

칸[0].bar(list(과목수.keys()), list(과목수.values()), color="tab:orange")
칸[0].set_title("좋아하는 과목")

칸[1].hist(수면, bins=5, color="tab:cyan", edgecolor="white")
칸[1].set_title("수면 시간 분포")
칸[1].set_xlabel("시간")

칸[2].scatter(통학, 수면, color="tab:purple")
칸[2].set_title("통학 시간과 수면 시간")
칸[2].set_xlabel("통학 시간 (분)")
칸[2].set_ylabel("수면 시간 (시간)")

plt.tight_layout()
plt.show()

# ⑤ 그래프를 보고 알게 된 점을 한 문장으로!
print("알게 된 점: 통학 시간이 길수록 ...")
```
