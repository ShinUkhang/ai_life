## 📈 1. 꺾은선 · 막대 · 비교 그래프

### 그래프 = 리스트 두 개

`plt.plot(x리스트, y리스트)`처럼 가로축과 세로축 리스트를 넣고, `plt.title()`·`plt.xlabel()`로 이름표를 붙인 뒤 `plt.show()`로 보여줘요.

### ① 꺾은선 그래프

```python
# ① 꺾은선 그래프 — 시간에 따라 어떻게 변할까?
import matplotlib.pyplot as plt

요일 = ["월", "화", "수", "목", "금", "토", "일"]
걸음수 = [6200, 7800, 5400, 9100, 8300, 12000, 4500]

plt.plot(요일, 걸음수, marker="o", color="tab:blue")   # x 리스트, y 리스트
plt.title("이번 주 걸음 수")
plt.xlabel("요일")
plt.ylabel("걸음 수")
plt.grid(True)
plt.show()

# 💡 두 리스트의 길이가 같아야 해요! 요일이 7개면 걸음수도 7개.
```

### ② 막대 · 가로 막대 그래프

```python
# ② 막대 그래프 — 무엇이 더 클까? (비교)
import matplotlib.pyplot as plt

과목 = ["국어", "영어", "수학", "과학", "사회"]
점수 = [85, 92, 78, 88, 95]
평균 = sum(점수) / len(점수)

plt.bar(과목, 점수, color="tab:orange")
plt.axhline(평균, color="gray", linestyle="--", label=f"평균 {평균:.1f}점")  # 가로 기준선
plt.title("과목별 점수")
plt.ylim(0, 100)
plt.legend()
plt.show()

# 가로 막대: 항목 이름이 길거나 많을 때 읽기 편해요
간식 = ["떡볶이", "마라탕", "탕후루", "붕어빵", "편의점 컵라면"]
표 = [12, 9, 5, 7, 10]

plt.barh(간식, 표, color="tab:green")
plt.title("우리 반이 좋아하는 간식")
plt.xlabel("표 수")
plt.show()
```

### ③ 비교 그래프 — 꺾은선 여러 개 · 묶은 막대

```python
# ③ 비교 그래프 — 두 그룹을 나란히 (꺾은선 여러 개 · 묶은 막대)
import matplotlib.pyplot as plt

월 = ["3월", "4월", "5월", "6월", "7월"]
우리반 = [72, 75, 78, 80, 83]
옆반 = [70, 74, 73, 79, 85]

# ① 꺾은선 여러 개: plot을 두 번 부르고 label + legend
plt.plot(월, 우리반, marker="o", label="우리 반")
plt.plot(월, 옆반, marker="s", label="옆 반")
plt.title("월별 수학 평균")
plt.ylabel("점수")
plt.legend()
plt.grid(True)
plt.show()

# ② 묶은 막대: 막대 위치를 왼쪽·오른쪽으로 조금씩 옮겨요
위치 = list(range(len(월)))          # [0, 1, 2, 3, 4]
왼쪽 = []
오른쪽 = []
for p in 위치:
    왼쪽.append(p - 0.2)
    오른쪽.append(p + 0.2)

plt.bar(왼쪽, 우리반, width=0.4, label="우리 반")
plt.bar(오른쪽, 옆반, width=0.4, label="옆 반")
plt.xticks(위치, 월)                 # 가로축 이름표를 월로
plt.title("월별 수학 평균 (묶은 막대)")
plt.ylim(0, 100)
plt.legend(loc="lower right")
plt.show()
```

## 🧮 3. 통계로 요약하기 — 대푯값 · 흩어짐 · 상관계수

### 숫자 하나로 데이터를 말하기

**대푯값**(평균 · 중앙값 · 최빈값)은 데이터의 '가운데', **흩어진 정도**(범위 · 표준편차 · 사분위수)는 데이터가 얼마나 퍼져 있는지 알려줘요. 파이썬에 들어 있는 `statistics` 모듈로 한 줄씩 계산해요.

### ⑧ 평균 · 중앙값 · 최빈값 · 표준편차 · 사분위수 · 이상치

```python
# ⑧ 통계로 요약하기 — 대푯값과 흩어진 정도
import statistics as st      # 파이썬에 들어 있는 통계 모듈

용돈 = [3, 5, 5, 4, 6, 5, 4, 7, 5, 50]     # 한 달 용돈(만 원), 마지막 한 명만 50!

print("평균:", st.mean(용돈))           # 모두 더해서 개수로 나누기
print("중앙값:", st.median(용돈))        # 크기순으로 줄 세웠을 때 가운데
print("최빈값:", st.mode(용돈))          # 가장 많이 나온 값
print()
print("범위:", max(용돈) - min(용돈))                 # 가장 큰 값 - 가장 작은 값
print("표준편차:", round(st.pstdev(용돈), 2))          # 평균에서 얼마나 흩어져 있나

Q1, Q2, Q3 = st.quantiles(용돈, n=4)                  # 사분위수 (4등분하는 세 값)
IQR = Q3 - Q1
print(f"Q1={Q1}, Q2(중앙값)={Q2}, Q3={Q3}, IQR={IQR}")

# 이상치 찾기: Q1 - 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 큰 값
이상치 = []
for 값 in 용돈:
    if 값 < Q1 - 1.5 * IQR or 값 > Q3 + 1.5 * IQR:
        이상치.append(값)
print("이상치:", 이상치)

# ❓ 평균(9.4)과 중앙값(5) 중 '보통 학생의 용돈'을 더 잘 나타내는 건?
#    이상치가 있을 때는 중앙값이 더 믿을 만해요!
```

### ⑨ 상관계수 — 관계의 세기를 숫자로

```python
# ⑨ 통계로 요약하기 — 상관계수: 두 값이 얼마나 함께 움직일까?
import statistics as st
import matplotlib.pyplot as plt

공부시간 = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
시험점수 = [55, 60, 58, 70, 72, 78, 80, 85, 83, 92]
게임시간 = [5, 4, 5, 3, 3, 2, 2, 1, 2, 1]

r1 = st.correlation(공부시간, 시험점수)
r2 = st.correlation(게임시간, 시험점수)
print(f"공부시간-점수 상관계수: {r1:.2f}")
print(f"게임시간-점수 상관계수: {r2:.2f}")

fig, 칸 = plt.subplots(1, 2, figsize=(11, 4))
칸[0].scatter(공부시간, 시험점수, color="tab:blue")
칸[0].set_title(f"공부 시간과 점수 (r = {r1:.2f})")
칸[0].set_xlabel("공부 시간")
칸[1].scatter(게임시간, 시험점수, color="tab:red")
칸[1].set_title(f"게임 시간과 점수 (r = {r2:.2f})")
칸[1].set_xlabel("게임 시간")
plt.tight_layout()
plt.show()

# 📏 상관계수 r 은 -1 ~ 1
#    1에 가까움: 같이 커짐(양의 상관) / -1에 가까움: 하나가 크면 하나는 작아짐(음의 상관)
#    0 근처: 거의 관계없음
# ⚠️ 상관관계 ≠ 인과관계! 아이스크림 판매량과 물놀이 사고는 함께 늘지만,
#    아이스크림이 사고를 일으키는 게 아니라 '더운 날씨'가 둘 다 늘린 거예요.
```

## 📄 4. CSV 파일 → 리스트 → 그래프

### CSV란?

**쉼표(,)로 칸을 나눈 표**예요. 엑셀·구글 시트·공공데이터포털(data.go.kr)·기상자료개방포털에서 CSV로 내려받을 수 있어요. 4일차에 배운 **파일 읽기**로 한 줄씩 꺼내 리스트에 담으면, 1~2에서 배운 그래프를 그대로 그릴 수 있어요.

⚠️ CSV에서 꺼낸 값은 모두 **문자열**이에요. 1일차 `input()`처럼 `float()`·`int()`로 숫자로 바꿔야 그래프가 돼요.

### 4-1. 연습용 CSV 파일 만들기

```python
# 4-1. 연습용 CSV 파일 만들기 (4일차에 배운 파일 쓰기!)
# CSV = 쉼표(,)로 칸을 나눈 표. 엑셀·구글 시트에서 '다운로드 → CSV'로 만들 수 있어요.

내용 = """월,평균기온,강수량
1월,-2.0,16.8
2월,0.7,28.2
3월,6.1,36.9
4월,12.6,72.9
5월,18.2,103.6
6월,22.7,129.5
7월,25.3,414.4
8월,26.1,348.2
9월,21.6,141.5
10월,14.8,52.2
11월,7.2,51.1
12월,0.4,22.6
"""

with open("서울_날씨.csv", "w", encoding="utf-8") as f:
    f.write(내용)

print("서울_날씨.csv 파일을 만들었어요! 왼쪽 📁 폴더 아이콘을 눌러 확인해 보세요.")
```

### 4-2. CSV 읽어서 리스트에 담기

```python
# 4-2. CSV 파일 읽어서 리스트에 담기
import csv

월 = []
기온 = []
강수량 = []

with open("서울_날씨.csv", encoding="utf-8") as f:
    읽기 = csv.reader(f)
    제목줄 = next(읽기)              # 첫 줄은 열 이름
    print("열 이름:", 제목줄)
    for 줄 in 읽기:                   # 줄 = ["1월", "-2.0", "16.8"]  ← 모두 문자열!
        월.append(줄[0])
        기온.append(float(줄[1]))     # 문자 → 숫자로 바꿔서 담기
        강수량.append(float(줄[2]))

print(월)
print(기온)

가장더운 = 기온.index(max(기온))     # 가장 큰 값이 몇 번째에 있는지
print(f"가장 더운 달: {월[가장더운]} ({기온[가장더운]}°C)")
print(f"1년 강수량 합계: {sum(강수량):.1f} mm")
```

### 4-3. 리스트로 그래프 그리기

```python
# 4-3. CSV에서 꺼낸 리스트로 그래프 그리기 (4-2를 먼저 실행하세요!)
import matplotlib.pyplot as plt

fig, 칸 = plt.subplots(1, 2, figsize=(13, 4))

칸[0].plot(월, 기온, marker="o", color="tab:red")
칸[0].set_title("서울 월별 평균 기온")
칸[0].set_ylabel("기온 (°C)")
칸[0].grid(True)

칸[1].bar(월, 강수량, color="tab:blue")
칸[1].set_title("서울 월별 강수량")
칸[1].set_ylabel("강수량 (mm)")

plt.tight_layout()
plt.show()

# ❓ 비가 가장 많이 오는 달은? 그때 기온은 어떤가요?
```

### 4-4. 내 CSV 파일 올리기 (코랩 전용)

## 🛠️ 종합 실습: 우리 반 설문 데이터 미니 보고서

CSV 만들기 → 리스트에 담기 → 딕셔너리로 세기 → 그래프 3개 → 알게 된 점까지, 데이터 과학의 6단계를 한 번에 해 봐요.

```python
# 🛠️ 종합 실습: 우리 반 설문 데이터 미니 보고서
import csv
import matplotlib.pyplot as plt

# ① 설문 결과 CSV 만들기 (실제 수업에서는 구글 설문 결과를 CSV로 받아 올려도 돼요)
내용 = """번호,통학시간,수면시간,좋아하는과목
1,15,7.0,수학
2,40,6.0,영어
3,25,6.5,과학
4,10,8.0,수학
5,55,5.5,국어
6,30,6.5,과학
7,20,7.5,수학
8,45,6.0,사회
9,35,6.5,영어
10,5,8.5,과학
11,50,5.0,수학
12,25,7.0,국어
"""
with open("설문.csv", "w", encoding="utf-8") as f:
    f.write(내용)

# ② 읽어서 리스트에 담기
통학 = []
수면 = []
과목들 = []
with open("설문.csv", encoding="utf-8") as f:
    읽기 = csv.reader(f)
    next(읽기)
    for 줄 in 읽기:
        통학.append(int(줄[1]))
        수면.append(float(줄[2]))
        과목들.append(줄[3])

# ③ 딕셔너리로 과목별 인원 세기 (3일차 복습!)
과목수 = {}
for 과목 in 과목들:
    과목수[과목] = 과목수.get(과목, 0) + 1
print("과목별 인원:", 과목수)

# ④ 그래프 3개로 보고서 만들기
fig, 칸 = plt.subplots(1, 3, figsize=(16, 4))

칸[0].bar(list(과목수.keys()), list(과목수.values()), color="tab:orange")
칸[0].set_title("좋아하는 과목")

칸[1].hist(수면, bins=5, color="tab:cyan", edgecolor="white")
칸[1].set_title("수면 시간 분포")
칸[1].set_xlabel("시간")

칸[2].scatter(통학, 수면, color="tab:purple")
칸[2].set_title("통학 시간과 수면 시간")
칸[2].set_xlabel("통학 시간 (분)")
칸[2].set_ylabel("수면 시간 (시간)")

plt.tight_layout()
plt.show()

# ⑤ 그래프를 보고 알게 된 점을 한 문장으로!
print("알게 된 점: 통학 시간이 길수록 ...")
```
