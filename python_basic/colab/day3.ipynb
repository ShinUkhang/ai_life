## 📦 1. 리스트 — 여러 개를 한 줄에 담기

### 리스트란?

장바구니에 여러 물건을 담듯이, **여러 데이터를 순서대로 담는 바구니**입니다. 대괄호 `[ ]`를 씁니다.

인덱스(번호)는 **0부터 시작**합니다! (첫 번째가 0번)

```python
# 리스트 만들기 (대괄호 사용!)
fruits = ["사과", "바나나", "딸기"]
print(fruits)          # 전체 출력

# 인덱싱: 번호로 하나씩 꺼내기 (0부터 시작!)
print(fruits[0])       # 사과 (첫 번째)
print(fruits[2])       # 딸기 (세 번째)
print(fruits[-1])      # 딸기 (뒤에서 첫 번째!)

# 슬라이싱: 구간으로 잘라내기 [시작:끝] (끝은 제외!)
print(fruits[0:2])     # 사과, 바나나

# append: 맨 뒤에 추가하기
fruits.append("포도")
print(fruits)          # 사과, 바나나, 딸기, 포도

# len: 개수 세기
print(len(fruits))     # 4개
```

### 튜플 — 못 바꾸는 리스트

소괄호 `( )`를 씁니다. 한 번 만들면 **바꿀 수 없습니다.** (좌표처럼 고정된 값에 사용)

```python
# 튜플 만들기 (소괄호 사용!)
point = (3, 5)         # x=3, y=5 좌표
print(point[0])        # 3
print(point[1])        # 5

# 튜플은 바꿀 수 없어요! (아래 줄을 실행하면 오류!)
# point[0] = 10        # ❌ TypeError 발생

# 리스트는 바꿀 수 있어요
nums = [1, 2, 3]
nums[0] = 100
print(nums)            # [100, 2, 3]
```

## 🗂️ 2. 딕셔너리 — 이름표가 붙은 서랍장

### 딕셔너리란?

리스트가 "0번, 1번" 번호로 찾는다면, 딕셔너리는 **"이름"이라는 키(key)로 값(value)을 찾습니다.**

중괄호 `{ }`와 콜론 `:`을 씁니다. (키: 값)

비유: **학생 성적표** — "철수"라는 이름으로 성적을 찾죠!

```python
# 딕셔너리 만들기 (중괄호 + 키: 값!)
scores = {"철수": 90, "영희": 85, "민수": 78}
print(scores)

# 키로 값 꺼내기
print(scores["철수"])    # 90
print(scores["영희"])    # 85

# 새로운 학생 추가하기
scores["지훈"] = 95
print(scores)

# 값 바꾸기
scores["민수"] = 80     # 민수 성적을 78 → 80으로 수정
print(scores)

# 키 목록, 값 목록 보기
print(scores.keys())    # 모든 이름
print(scores.values())  # 모든 점수
```

### 집합 — 중복 없는 주머니

중괄호 `{ }`를 쓰지만 키 없이 값만 담고, **중복을 허용하지 않습니다.**

```python
# 집합 만들기
my_set = {1, 2, 3, 3, 2, 1}
print(my_set)           # {1, 2, 3} → 중복 제거!

# 추가하기
my_set.add(4)
print(my_set)           # {1, 2, 3, 4}

# 합집합·교집합
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)            # 합집합: {1, 2, 3, 4, 5}
print(a & b)            # 교집합: {3}
```

## ⚙️ 3. 함수 — 자주 쓰는 코드를 이름 붙이기

### 함수란?

자주 쓰는 코드를 **하나의 이름으로 묶어두는 것**입니다. `def`로 만듭니다.

비유: **자판기** — 돈(입력)을 넣으면 음료수(출력)가 나옵니다!

- **매개변수**: 자판기에 넣는 돈 (받는 쪽 이름)

- **인자**: 실제로 넣는 돈 (호출할 때 값)

- **return**: 자판기에서 나오는 음료수 (돌려주는 값)

```python
# 함수 정의하기 (def 사용!)
def 인사하기():
    print("안녕하세요!")    # ← 들여쓰기 필수!

# 함수 호출하기 (이름 + 괄호)
인사하기()                # 안녕하세요!
인사하기()                # 또 호출 가능!

# 매개변수가 있는 함수
def 인사하기_이름(name):   # name = 매개변수 (받는 쪽)
    print(name + "님, 안녕하세요!")

인사하기_이름("철수")      # "철수" = 인자 (주는 쪽)
인사하기_이름("영희")

# return: 결과를 돌려주는 함수
def 더하기(a, b):
    결과 = a + b
    return 결과          # ← 이 값을 돌려줍니다!

합 = 더하기(3, 5)          # 8이 돌아와서 합에 저장
print(합)

# return이 없으면? → None (없음)이 돌아와요
def 그냥출력():
    print("출력만!")

x = 그냥출력()
print(x)                   # None
```

## 🎯 종합 실습: 성적 관리 프로그램

지금까지 배운 **딕셔너리 + 함수**를 합쳐서 성적 관리 프로그램을 만들어봅시다!

```python
# ===== 성적 관리 프로그램 =====
# 딕셔너리: 학생 이름 → 점수 저장
scores = {}

# 학생 추가 함수
def 학생추가(이름, 점수):
    scores[이름] = 점수
    print(이름 + " 학생 추가 완료! (점수: " + str(점수) + ")")

# 평균 구하는 함수
def 평균구하기():
    if len(scores) == 0:       # 학생이 없으면
        return 0
    합계 = sum(scores.values())  # 모든 점수 더하기
    return 합계 / len(scores)    # 평균 = 합계 ÷ 인원수

# 최고 점수 학생 찾기 함수
def 최고점수():
    if len(scores) == 0:
        return "학생이 없습니다"
    최고이름 = max(scores, key=scores.get)  # 점수가 가장 높은 이름
    return 최고이름 + " (" + str(scores[최고이름]) + "점)"

# ===== 사용해보기 =====
학생추가("철수", 90)
학생추가("영희", 85)
학생추가("민수", 78)
학생추가("지훈", 95)

print("\n📊 전체 성적:", scores)
print("📊 평균:", 평균구하기())
print("🏆 1등:", 최고점수())
```
