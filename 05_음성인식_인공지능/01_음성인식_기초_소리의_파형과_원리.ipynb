## 1. ⚙️ 준비 운동: 필요한 파이썬 도구 상자 불러오기
소리를 만들고, 눈으로 그려보고, 귀로 들어보기 위해 파이썬의 대표 도구들을 불러옵니다.
* `numpy`: 숫자를 계산하고 소리 데이터를 만드는 수학 마법사
* `matplotlib`: 소리를 예쁜 그래프로 그려주는 미술가
* `IPython.display.Audio`: 노트북 안에서 소리를 직접 들려주는 음악 플레이어
* `scipy`: 소리 신호를 분석해주는 과학자

> ✅ **체크포인트**: `🎉 준비 완료!` 메시지가 떴다 → 성공! 이제 소리 실험실 문이 열렸어요.


```python
# 1. 필수 라이브러리 불러오기
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
from scipy import signal

# 한글 폰트 및 그래프 스타일 깔끔하게 설정
plt.rcParams['figure.figsize'] = (10, 4)
plt.rcParams['font.size'] = 12
print("🎉 준비 완료! 이제 소리를 만들어볼까요?")
```

---
## 2. 🌊 소리란 무엇일까요? (공기의 떨림 ➔ 사인파)
소리는 **'공기의 진동(떨림)'**입니다.
우리가 목소리를 내면 목의 성대가 떨리고, 그 떨림이 공기를 타고 상대방의 귓속 고막을 때립니다.

컴퓨터는 이 떨림을 수학 공식인 **사인파(Sine Wave, $\sin$)** 곡선으로 이해합니다. 사인파는 물결처럼 출렁이는 가장 기본적인 소리의 모양이에요.
* **진폭 (Amplitude)**: 물결의 높이 ➔ **소리의 크기(볼륨)** (물결이 높을수록 큰 소리, 낮을수록 작은 소리)
* **주파수 (Frequency, Hz)**: 1초 동안 물결이 몇 번 출렁이는가? ➔ **소리의 높낮이(음정)** (숫자가 크면 뾰족한 고음, 작으면 묵직한 저음)


```python
# 2. 피아노 '라(A4 - 440Hz)' 소리를 만들어봅시다!
sampling_rate = 16000  # 1초에 16,000번 소리를 잘게 쪼개어 숫자로 기록 (샘플링: 소리를 숫자로 바꾸는 과정)
duration = 2.0         # 2초 동안 소리 재생
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)

# 주파수 440Hz (음악에서 모든 악기를 조율하는 기준 음 '라')
frequency = 440
sound_wave = 0.5 * np.sin(2 * np.pi * frequency * t)

print("🎵 440Hz '라' 소리가 만들어졌습니다! 아래 재생 버튼을 눌러보세요.")
ipd.Audio(sound_wave, rate=sampling_rate)
```

### 🔍 소리의 물결을 눈으로 확인해볼까요?
방금 들은 소리의 아주 일부분(0.01초)을 현미경처럼 확대해서 그려봅시다!

> ✅ **체크포인트**: 파란 물결 그래프가 그려졌다 → 성공! 이 물결이 바로 소리의 얼굴(파형)이에요.


```python
# 0.01초 동안의 소리 파형(Waveform) 확대해서 그리기
plt.figure(figsize=(10, 3))
plt.plot(t[:160], sound_wave[:160], color='#4F46E5', linewidth=2)
plt.title("Waveform of 440Hz Sound (0.01 sec)", fontweight='bold')
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (Air vibration)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. 🎮 [미니 미션] 모기 소리 vs 덤프트럭 소리 만들기!
주파수(Hz) 숫자를 바꿔가며 소리의 높낮이를 체험해봅시다.
* 🐻 **덤프트럭 저음**: 100Hz (낮고 웅장한 소리)
* 🎹 **피아노 도(C4)**: 261Hz
* 🦟 **짜증 나는 모기 소리**: 1,200Hz ~ 2,000Hz (아주 얇고 뾰족한 고음)

> ✅ **체크포인트**: `my_frequency` 숫자를 바꿀 때마다 소리의 높낮이가 달라진다 → 성공! 이제 여러분은 주파수 DJ예요. 🎧


```python
def make_beep(freq, seconds=1.5):
    t_test = np.linspace(0, seconds, int(sampling_rate * seconds), endpoint=False)
    wave = 0.5 * np.sin(2 * np.pi * freq * t_test)
    return ipd.Audio(wave, rate=sampling_rate)

# 👇 아래 숫자를 자유롭게 바꿔보세요! (예: 120, 261, 880, 1500)
my_frequency = 880 
print(f"🔊 현재 주파수: {my_frequency} Hz")
make_beep(my_frequency)
```

## 4. 🎹 코드로 연주하는 '도-미-솔' 화음 만들기!
여러 개의 소리 파형을 수학적으로 더하면($wave_1 + wave_2 + wave_3$) 아름다운 **화음(Chord)**이 탄생합니다.

> ✅ **체크포인트**: 도-미-솔이 동시에 울리는 화음이 들렸다 → 성공! 파형을 더하는 게 곧 소리를 섞는 거예요.


```python
# 도(C4), 미(E4), 솔(G4) 주파수
freq_do = 261.63
freq_mi = 329.63
freq_sol = 392.00

wave_do = np.sin(2 * np.pi * freq_do * t)
wave_mi = np.sin(2 * np.pi * freq_mi * t)
wave_sol = np.sin(2 * np.pi * freq_sol * t)

# 세 파형을 더하면 C 코드(C 메이저) 화음이 됩니다!
chord = (wave_do + wave_mi + wave_sol) / 3

print("✨ 아름다운 C 메이저 화음 (도-미-솔) 듣기:")
ipd.Audio(chord, rate=sampling_rate)
```

```python
# 화음의 파형은 단일 음과 달리 복잡하고 풍성한 모양을 가집니다!
plt.figure(figsize=(10, 3))
plt.plot(t[:300], chord[:300], color='#10B981', linewidth=2)
plt.title("Waveform of C-Major Chord (Do-Mi-Sol)", fontweight='bold')
plt.xlabel("Time (seconds)")
plt.ylabel("Combined Amplitude")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. 🌈 인공지능의 비밀 무기: 스펙트로그램(Spectrogram)
그런데 문제가 있습니다!
컴퓨터에게 위의 파형 그래프만 보여주면 **"이게 사람이 말한 건지, 개가 짖은 건지, '안녕'이라고 한 건지"** 도무지 알 수가 없습니다. 🤯

그래서 인공지능 과학자들은 소리를 **무지개 그림(스펙트로그램)**으로 바꿉니다!
* **가로축**: 시간 (소리가 흘러가는 시간)
* **세로축**: 주파수 (어떤 높낮이의 음들이 섞여 있는가?)
* **색깔의 밝기**: 에너지(세기) (밝을수록 그 소리가 강력하게 울리는 중!)

> 💡 **푸리에 변환이 뭐예요?** 복잡하게 뒤엉킨 소리 파형을 "어떤 높낮이의 음이 얼마나 섞였는지"로 분해해주는 수학 도구예요. 스펙트로그램은 이 분해를 시간의 흐름에 따라 나열한 그림이랍니다. 이름은 어렵지만, 하는 일은 "소리 재료 분리"예요! 🔬

> **인공지능(AI)은 음성인식을 할 때 소리를 듣는 것이 아니라, 바로 이 스펙트로그램 '그림'을 눈으로 보고 글자로 읽어내는 것입니다!** 🖼️👀

> ✅ **체크포인트**: 무지개 그림 속에 도·미·솔 세 줄이 선명하게 보였다 → 성공! 여러분은 이제 AI의 눈으로 소리를 봤어요.


```python
# 피아노 화음의 스펙트로그램을 그려봅시다!
frequencies, times, spectrogram = signal.spectrogram(chord, sampling_rate)

plt.figure(figsize=(10, 4))
plt.pcolormesh(times, frequencies, 10 * np.log10(spectrogram + 1e-10), shading='gouraud', cmap='viridis')
plt.title("Spectrogram: Time vs Frequency (How AI Sees Sound!)", fontweight='bold')
plt.ylabel('Frequency [Hz]')
plt.xlabel('Time [sec]')
plt.ylim(0, 1000) # 사람이 주로 듣는 1000Hz 이하 확대
plt.colorbar(label='Power/Frequency (dB)')
plt.axhline(y=freq_do, color='r', linestyle='--', alpha=0.7, label='Do (261Hz)')
plt.axhline(y=freq_mi, color='y', linestyle='--', alpha=0.7, label='Mi (330Hz)')
plt.axhline(y=freq_sol, color='cyan', linestyle='--', alpha=0.7, label='Sol (392Hz)')
plt.legend(loc='upper right')
plt.show()
print("💡 세 개의 가로선(도, 미, 솔)이 선명하게 보이시나요? AI는 이 줄무늬를 보고 무슨 소리인지 알아챕니다!")
```
